```python
import random

board = [' ' for _ in range(9)]

def display():
    print()
    print(board[0], '|', board[1], '|', board[2])
    print('--+---+--')
    print(board[3], '|', board[4], '|', board[5])
    print('--+---+--')
    print(board[6], '|', board[7], '|', board[8])
    print()

def check_winner(p):
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] == p:
            return True
    return False

def ai_move():
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            if check_winner('O'):
                return
            board[i] = ' '

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'
            if check_winner('X'):
                board[i] = ' '
                board[i] = 'O'
                return
            board[i] = ' '

    empty = [i for i in range(9) if board[i] == ' ']
    if empty:
        board[random.choice(empty)] = 'O'

def game():
    for turn in range(9):
        display()

        if turn % 2 == 0:
            pos = int(input("Player X, enter position (1-9): ")) - 1

            if pos < 0 or pos > 8 or board[pos] != ' ':
                print("Invalid position!")
                continue

            board[pos] = 'X'

            if check_winner('X'):
                display()
                print("Player X wins!")
                return

        else:
            print("AI is thinking...")
            ai_move()

            if check_winner('O'):
                display()
                print("AI wins!")
                return

    display()
    print("Game Draw!")

game()
```

```python
def simple_reflex_agent(status, position, n):
    if status == "Dirty":
        return "Suck"
    elif position < n - 1:
        return "Move Right"
    else:
        return "Move Left"


def goal_based_agent(rooms, position):
    actions = []

    while "Dirty" in rooms:
        if rooms[position] == "Dirty":
            actions.append(f"Cleaning Room {position + 1}")
            rooms[position] = "Clean"

        elif position < len(rooms) - 1:
            position += 1
            actions.append(f"Move to Room {position + 1}")

        else:
            position -= 1
            actions.append(f"Move to Room {position + 1}")

    return actions


n = int(input("Enter number of rooms: "))

rooms = []
for i in range(n):
    status = input(f"Enter status of Room {i + 1} (Clean/Dirty): ")
    rooms.append(status)

position = int(input("Enter starting room: ")) - 1

print("\n--- Simple Reflex Agent ---")

action = simple_reflex_agent(
    rooms[position], position, n
)

print("Action:", action)

print("\n--- Goal Based Agent ---")

actions = goal_based_agent(rooms, position)

for action in actions:
    print(action)

print("\nFinal State:")
for i in range(n):
    print(f"Room {i + 1}: {rooms[i]}")

print("\nGoal Achieved: All rooms are clean")
```
